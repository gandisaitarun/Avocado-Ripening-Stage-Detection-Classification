# 04 — Held-out evaluation and single-image demo

Evaluate the saved classical and deep-learning models exactly once on the supplied test set, show the metric comparison and confusion matrices, then run both predictors on one example image. Do not tune a model based on this test-set result.

```python
from pathlib import Path
import os
root = Path.cwd()
while not (root / "src").is_dir() and root != root.parent:
    root = root.parent
if not (root / "src").is_dir():
    raise FileNotFoundError("Open this notebook from the project checkout.")
os.chdir(root)
print("Project:", root.resolve())
```

```python
import subprocess, sys

subprocess.run([
    sys.executable, "-m", "scripts.evaluate_models",
    "--test-dir", "test",
    "--classical-model", "models/classical_svm.joblib",
    "--deep-model", "models/mobilenet_v2.pth",
    "--output-dir", "report/results",
], check=True)
```

```python
import json
import pandas as pd

with open("report/results/model_comparison.json", encoding="utf-8") as file:
    results = json.load(file)
comparison = pd.DataFrame(results["metrics"]).T[["accuracy", "macro_precision", "macro_recall", "macro_f1"]]
display(comparison.style.format("{:.2%}"))
print("Held-out test images:", results["test_image_count"])
print("Class order:", results["class_names"])
```

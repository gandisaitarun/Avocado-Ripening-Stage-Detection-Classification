```python
from pathlib import Path
import os
from IPython.display import display, Image as DisplayImage

root = Path.cwd()
while not (root / "src").is_dir() and root != root.parent:
    root = root.parent
if not (root / "src").is_dir():
    raise FileNotFoundError("Open this notebook from the project checkout.")
os.chdir(root)
print("Project:", root.resolve())
```

```python
from src.dataset import discover_images
from collections import Counter

train_paths, train_labels, class_names = discover_images(Path("train"))
test_paths, test_labels, test_classes = discover_images(Path("test"))
if class_names != test_classes:
    raise ValueError(f"Train/test classes differ: {class_names} vs {test_classes}")
print("Classes:", class_names)
print("Train:", len(train_paths), Counter(train_labels))
print("Test:", len(test_paths), Counter(test_labels))
```

# 02 — Classical computer vision baseline

Build normalized HSV and CIELAB color histograms plus uniform LBP texture histograms, fit a class-balanced RBF SVM on the training partition, and report validation metrics. The external test folder is not used for fitting or model selection.

```python
from pathlib import Path
import os
root = Path.cwd()
while not (root / "src").is_dir() and root != root.parent:
    root = root.parent
if not (root / "src").is_dir():
    raise FileNotFoundError("Open this notebook from the project checkout.")
os.chdir(root)
print("Project:", root.resolve())
```

```python
from src.dataset import discover_images, stratified_train_validation_split
from src.classical_model import train_classical_model, save_classical_model, extract_feature_matrix
from sklearn.metrics import classification_report, accuracy_score

paths, labels, class_names = discover_images(Path("train"))
split = stratified_train_validation_split(paths, labels, random_state=42)
train_paths, train_labels = split["train"]
val_paths, val_labels = split["validation"]
print(f"Fit images: {len(train_paths)}; validation images: {len(val_paths)}")
print("Class order:", class_names)
```

```python
model = train_classical_model(train_paths, train_labels, random_state=42)
val_predictions = model.predict(extract_feature_matrix(val_paths))
print("Validation accuracy:", accuracy_score(val_labels, val_predictions))
print(classification_report(val_labels, val_predictions, labels=class_names, zero_division=0))
save_classical_model(model, class_names, Path("models/classical_svm.joblib"))
```
